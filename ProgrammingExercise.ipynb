```python
class Solution:
    def sort012(self, arr):
        low, mid, high = 0, 0, len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
            elif arr[mid] == 1:
                mid += 1
            else:
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1
```

```python
class Solution:
    def firstSearch(self, arr, k):
        low, high = 0, len(arr) - 1
        ans = -1
        
        while low <= high:
            mid = low + (high - low) // 2
            
            if arr[mid] == k:
                ans = mid
                high = mid - 1
            elif arr[mid] < k:
                low = mid + 1
            else:
                high = mid - 1
                
        return ans
```
